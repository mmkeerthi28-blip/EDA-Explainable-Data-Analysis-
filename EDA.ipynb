DATA VISUALIZATION

```python
###Data visualization aims to represent data in a visual context to make it easier to understand, highlight important aspects, and communicate findingd effectively to stakeholders.

### . Bar Chart   syntax for all from seaborn package(sns)  :   sns.barplot(x='colname',y='colname',data=df)
### . Line Chart                               :   sns.lineplot(x='colname',y='colname',data=df)
### . Scatter                                  :   sns.hisplot(df['colname'],bins=10,kde=False)
### . Histogram                                :   sns.scatterplot(x='colname',y='colname',data=df)

### barchart used : one categorical value or one numerical value
### linechart used : continous value ex:month wise sales
### scatter used : relationship (between two numerical columns etc)
### Histogram used : analyze a single column(how myvalues are distributed)
```

```python
plt.figure(figsize=(8,6))
sns.barplot(x='Sex',y='Survived',data=df,color='pink',edgecolor='red')
plt.title("Gender vs Survival Rate") ### id i want title i can give by this way , xlabel give x axis name
plt.xlabel("Gender")
plt.ylabel("Survived")
plt.show()
```

```python
#age distribution lets see

plt.figure(figsize=(8,6))
sns.histplot(df['Age'],bins=20,color='teal')
```

```python
plt.figure(figsize=(8,6))
sns.histplot(df['Age'],bins=30,color='teal',kde=True)
```

```python
#scatter
sns.scatterplot(x='Age',y='Fare',data=df)
```

```python

sns.scatterplot(x='Age',y='Fare',data=df,hue=df['Pclass'],palette='viridis')# palette for color
plt.title('Relatioship btw Fare and Age')
```

```python
#linechart only passenger id is continous so we can do for that
sns.lineplot(x="PassengerId",y='Fare',data=df)
```
